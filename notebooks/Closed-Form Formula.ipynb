```python
import numpy as np
from scipy.stats import norm

def closed_form_formula(S0,K,T,r,sigma,t):  
    if T-t <= 0:
        return max(S0-K,0)
    d1 = (np.log(S0/K)+(r+0.5*sigma**2)*(T-t))/(sigma*np.sqrt(T-t))
    d2 = d1-sigma*np.sqrt(T-t)
    C = S0*norm.cdf(d1)-np.exp(-r*(T-t))*K*norm.cdf(d2)
    return C
```
